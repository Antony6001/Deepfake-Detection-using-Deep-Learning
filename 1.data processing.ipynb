Build a deep learning system that classifies human face images as REAL or FAKE, while displaying the available gender and age-group information.

**Dataset → Download Images → EDA → Preprocessing → Augmentation → Model Training → Evaluation → Grad-CAM → Prediction → Deployment**

```python
import os
import time
import requests
import pandas as pd
from tqdm import tqdm
```

```python
df = pd.read_csv("dataset.csv")
os.makedirs("data/images", exist_ok=True)

for _, row in tqdm(df.iterrows(), total=len(df)):
    path = f"data/images/{row['image_id']}.jpg"

    try:
        r = requests.get(row["image_url"], timeout=15)
        r.raise_for_status()

        with open(path, "wb") as f:
            f.write(r.content)

    except Exception as e:
        print("Failed:", row["image_id"], e)
```

```python


df = pd.read_csv("dataset.csv")

os.makedirs("data/images", exist_ok=True)

session = requests.Session()

MAX_RETRIES = 3
TIMEOUT = 15

failed_images = []

for _, row in tqdm(df.iterrows(), total=len(df)):

    image_id = row["image_id"]
    url = row["image_url"]

    path = f"data/images/{image_id}.jpg"

    # Already downloaded
    if os.path.exists(path) and os.path.getsize(path) > 0:
        continue

    success = False

    for attempt in range(1, MAX_RETRIES + 1):

        try:

            response = session.get(
                url,
                timeout=TIMEOUT
            )

            # 403 - Forbidden
            if response.status_code == 403:
                print(f"\n403 Forbidden: {image_id}")
                break

            # 404 - Not Found
            if response.status_code == 404:
                print(f"\n404 Not Found: {image_id}")
                break

            # 429 - Too many requests
            if response.status_code == 429:
                print(f"\n429 Rate Limited: {image_id}")
                time.sleep(10)
                continue

            response.raise_for_status()

            with open(path, "wb") as f:
                f.write(response.content)

            success = True
            break

        except requests.exceptions.Timeout:

            print(
                f"\nTimeout {attempt}/{MAX_RETRIES}: "
                f"{image_id}"
            )

            time.sleep(2)

        except requests.exceptions.ConnectionError:

            print(
                f"\nConnection error {attempt}/{MAX_RETRIES}: "
                f"{image_id}"
            )

            time.sleep(3)

        except Exception as e:

            print(
                f"\nError for {image_id}: {e}"
            )

            break

    if not success:

        failed_images.append({
            "image_id": image_id,
            "image_url": url
        })


# Save failed images
failed_df = pd.DataFrame(failed_images)

failed_df.to_csv(
    "failed_images.csv",
    index=False
)

print("\n==============================")
print("Download completed")
print("==============================")
print("Total images :", len(df))
print("Failed images:", len(failed_df))
print("Failed list  : failed_images.csv")
```
