## Basic EDA

```python
#Required Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('dataset_clean.csv')
```

```python
df.head()
```

**Real and Fake Image Distribution**

```python
sns.countplot(data=df, x='label')

plt.title("Real vs Fake Image Distribution")
plt.xlabel("Label")
plt.ylabel("Count")
plt.show()
```

**Gender Distribution**

```python
sns.countplot(data=df, x='gender')

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

**Age Group Distribution**

```python
sns.countplot(data=df, x='age_group')

plt.title("Age Group Distribution")
plt.xlabel("Age Group")
plt.ylabel("Count")
plt.show()
```

**Label Distribution by Gender**

```python
sns.countplot(data=df, x='gender', hue='label')

plt.title("Label Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

**Label Distribution by Age Group**

```python
sns.countplot(data=df, x='age_group', hue='label')

plt.title("Label Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Count")
plt.show()
```

**Gender Distribution Across Age Groups**

```python
sns.countplot(data=df, x='age_group',hue='gender')

plt.title("Gender Distribution Across Age Groups")
plt.xlabel("Age Group")
plt.ylabel("Count")
plt.show()
```
